# 4. Varsayılan parametreler ve alternatif constructor

Python klasik metot aşırı yüklemeyi desteklemez. Bunun yerine varsayılan parametreler veya
anlamlı isimlere sahip `classmethod` constructor'lar kullanılır.

```python
class Product:
    def __init__(self, name: str, price: float = 0, quantity: int = 1) -> None:
        self.name = name
        self.price = price
        self.quantity = quantity

    @classmethod
    def from_csv_row(cls, row: str) -> "Product":
        name, price, quantity = (value.strip() for value in row.split(","))
        return cls(name, float(price), int(quantity))

    def __repr__(self) -> str:
        return f"Product(name={self.name!r}, price={self.price}, quantity={self.quantity})"
```

```python
default_stock = Product("Fare")
imported = Product.from_csv_row("Klavye, 4999.99, 5")

print(default_stock)
print(imported)
```
