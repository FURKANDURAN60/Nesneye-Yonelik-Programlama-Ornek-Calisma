# 5. Kalıtım ve metot ezme

Alt sınıflar ortak davranışı temel sınıftan alır, gerektiğinde metotları kendi ihtiyaçlarına göre ezer.

```python
class Product:
    def __init__(self, name: str, price: float) -> None:
        self.name = name
        self.price = price

    def delivery_message(self) -> str:
        return "Standart teslimat uygulanır."


class PhysicalProduct(Product):
    def __init__(self, name: str, price: float, weight_kg: float) -> None:
        super().__init__(name, price)
        self.weight_kg = weight_kg

    def delivery_message(self) -> str:
        return f"{self.weight_kg} kg ürün kargoya verildi."


class DigitalProduct(Product):
    def __init__(self, name: str, price: float, download_url: str) -> None:
        super().__init__(name, price)
        self.download_url = download_url

    def delivery_message(self) -> str:
        return f"İndirme bağlantısı: {self.download_url}"
```

```python
products = [
    PhysicalProduct("Klavye", 4_999.99, 0.8),
    DigitalProduct("Python Eğitimi", 899.99, "https://example.com/download"),
]

for product in products:
    print(product.name, "-", product.delivery_message())
```
