# 2. Yapıcı metotlar

`__init__`, yeni bir nesnenin başlangıç durumunu tek yerde tanımlar.

```python
class Product:
    def __init__(self, name: str, price: float, quantity: int) -> None:
        self.name = name
        self.price = price
        self.quantity = quantity

    def total_price(self) -> float:
        return self.price * self.quantity
```

```python
laptop = Product("Laptop", 49_999.99, 3)
phone = Product("Akıllı Telefon", 29_999.99, 5)

print(laptop.name, laptop.total_price())
print(phone.name, phone.total_price())
```
