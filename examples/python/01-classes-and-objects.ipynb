# 1. Sınıflar ve nesneler

Sınıf bir şablondur; nesne ise bu şablondan üretilen bağımsız bir örnektir.

```python
class Product:
    category = "Elektronik"


laptop = Product()
laptop.name = "Laptop"
laptop.price = 49_999.99

phone = Product()
phone.name = "Akıllı Telefon"
phone.price = 29_999.99
```

```python
print(laptop.name, laptop.price, laptop.category)
print(phone.name, phone.price, phone.category)
print(laptop is phone)
```
