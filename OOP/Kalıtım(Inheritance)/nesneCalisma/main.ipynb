```python
class Person:
    
    def __init__(self, name=None, age=None):
        self._name = name
        self._age = age
        print("Person Cretated")
    
        if self._name is not None and self._age is not None:
            print("Person created with parameters")
    
    @property
    def name(self):
        return self._name
    @property
    def age(self):
        return self._age
    @name.setter
    def name(self, value):
        self._name = value
    @age.setter
    def age(self, value):
        self._age = value
    
    def ToString(self):
        return f"Name: {self._name}, Age: {self._age}"

class Teacher(Person):
    
    def __init__(self, name, age, subject=None):
        super().__init__(name, age)
        
        self._subject = subject
        
        print(f"Teacher created with parameters and Teacher subject :  {self.subject}")

    @property
    def subject(self):
        return self._subject
    
    @subject.setter
    def subject(self, value):
        self._subject = value
    
    def ToString(self):
        return f"Name: {self._name}, Age: {self._age}, Subject: {self._subject}"
        
class Student(Person):
    
    def __init__(self, name, age, studentNumber=None):
        super().__init__(name, age)
        
        self._studentNumber = studentNumber
        
        print(f"Student created with parameters and Student number :  {studentNumber}")
        
    
    @property
    def studentNumber(self):
        return self._studentNumber
    @studentNumber.setter
    def studentNumber(self, value):
        self._studentNumber = value
        
    def ToString(self):
        return f"Name: {self._name}, Age: {self._age}, Student Number: {self._studentNumber}"


teacher1 = Teacher("Halit Öztekin", 45, "Computer Architecture")

print(teacher1.ToString())

student1 = Student("Furkan", 23, "B210109048")
print(student1.ToString())
        
```
