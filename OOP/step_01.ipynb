```python
class Product:
    def __init__(self):
        print(f"An instance has been derived from the Product class.")
        pass
    pass
```

```python
item1 = Product()

#Dinamik özelik tanımı
item1.name = "Laptop"
item1.price = 49999.99
item1.quantity = 3
```

```python
print(item1)
print(item1.name)
print(item1.price)
print(item1.quantity)
```

```python
print(type(item1))
print(type(item1.name))
print(type(item1.price))
print(type(item1.quantity))
```
